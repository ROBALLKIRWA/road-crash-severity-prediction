### **Understand these three terms**

**Accuracy:** How many predictions were correct out of all predictions.

**Precision:** When the model predicts class 1, how often is it correct?

**Recall:** Out of all the actual class 1 records, how many did the model successfully identify?

For your project, recall is particularly important because missing a severe crash category could be costly.
